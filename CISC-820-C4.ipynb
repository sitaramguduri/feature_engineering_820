```python
import numpy as np


def least_squares(z, y):
    z = np.array(z, dtype=float)
    y = np.array(y, dtype=float)

    n, d = z.shape
    M = np.zeros((d, d))
    s = np.zeros(d)

    for i in range(n):
        M += np.outer(z[i], z[i])
        s += z[i] * y[i]
    w = numpy.linalg.solve(M,s)
    R = 0

    for i in range(n):
        prediction = w @ z[i]
        R += (prediction - y[i]) ** 2

    return w, R


```

```python
x = [0.23, 0.88, 0.21, 0.92, 0.49, 0.62, 0.77, 0.52, 0.30, 0.19]
y = [0.19, 0.96, 0.33, 0.80, 0.46, 0.45, 0.67, 0.32, 0.38, 0.37]
z = []

for xi in x:
    z.append([xi, 1, xi**2])


w, R = least_squares(z, y)
a = w[0]
b = w[1]
c = w[2]

plot_fit(
    x,
    y,
    lambda x: a*x + b + c*x**2
)
```
